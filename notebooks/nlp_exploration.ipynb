# NLP Exploration Notebook
This notebook walks through the core NLP tasks implemented in this project:
1. Text Preprocessing
2. Text Classification (Sentiment Analysis)
3. Named Entity Recognition (NER)
4. Word Frequency Visualization

```python
import sys
sys.path.insert(0, '..')

import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud

from src.preprocessing import TextPreprocessor, get_word_frequencies
from src.classification import TextClassifier, train_sentiment_demo
from src.ner import spacy_ner, highlight_entities

print('All imports successful!')
```

## 1. Text Preprocessing

```python
preprocessor = TextPreprocessor(
    lowercase=True,
    remove_stopwords=True,
    remove_punctuation=True,
    lemmatization=True,
)

texts = [
    "Natural Language Processing enables computers to understand human language.",
    "Machine learning models are trained on large datasets to learn patterns.",
    "Deep learning has revolutionized NLP tasks like translation and summarization.",
]

for text in texts:
    tokens = preprocessor.process(text)
    print(f'Original : {text}')
    print(f'Tokens   : {tokens}')
    print()
```

## 2. Word Frequency & Word Cloud

```python
corpus = ' '.join(texts)
all_tokens = preprocessor.process(corpus)
freq = get_word_frequencies(all_tokens)

# Bar chart — top 15 words
top_n = dict(list(freq.items())[:15])
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].barh(list(top_n.keys())[::-1], list(top_n.values())[::-1], color='steelblue')
axes[0].set_title('Top 15 Word Frequencies')
axes[0].set_xlabel('Count')

# Word cloud
wc = WordCloud(width=600, height=400, background_color='white').generate_from_frequencies(freq)
axes[1].imshow(wc, interpolation='bilinear')
axes[1].axis('off')
axes[1].set_title('Word Cloud')

plt.tight_layout()
plt.show()
```

## 3. Text Classification (Sentiment Analysis)

```python
print('Training sentiment classifier...')
clf = train_sentiment_demo()

test_texts = [
    "This product is absolutely fantastic!",
    "Terrible quality, never buying again.",
    "It's okay, does what it says.",
    "Exceeded all my expectations!",
    "Complete waste of money.",
]

preds = clf.predict(test_texts)
probs = clf.predict_proba(test_texts)

results_df = pd.DataFrame({
    'Text': test_texts,
    'Prediction': preds,
    'Confidence': [f'{max(p):.1%}' for p in probs],
})
display(results_df)
```

## 4. Named Entity Recognition

```python
ner_texts = [
    "Apple was founded by Steve Jobs in Cupertino, California.",
    "The UN headquarters is in New York City.",
    "Elon Musk leads Tesla and SpaceX.",
]

results = spacy_ner(ner_texts)

for text, entities in zip(ner_texts, results):
    print(f'Text: {text}')
    if entities:
        print(f'Highlighted: {highlight_entities(text, entities)}')
        for ent in entities:
            print(f'  [{ent["label"]}] {ent["text"]} — {ent["description"]}')
    print()
```
